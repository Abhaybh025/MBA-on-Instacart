## Data Preparation 

For Analysis, I am combining orders_product_train and order_product_prior as order_products. Based on order_id, product_id, aisle_id and department_id I will merge those data with order_products and will create one single dataframe.

```python
order_products = pd.concat([order_products_prior, order_products_train], ignore_index=True)
order_products.shape
```

```python
order_products = order_products.merge(products, on='product_id', how='left')
order_products = order_products.merge(aisles, on='aisle_id', how='left')
order_products = order_products.merge(departments, on='department_id', how='left')
order_products = order_products.merge(orders, on='order_id', how='left')
order_products.shape
```

### Reducing memory usage for faster analysis 

```python
def reduce_memory(df):
    """
    This function reduces the dataframe memory usage by converting its type for easier handling.
    Parameters: Dataframe
    Return: Dataframe
    """

    initial_mem_usage = df.memory_usage().sum()/1024**2
    print('memory usage of properties dataframe is:', initial_mem_usage,'MB')

    for col in df.columns:
        if df[col].dtype in ["int64","int32","int16"]:

            cmin = df[col].min()
            cmax = df[col].max()

            if cmin > np.iinfo(np.int8).min and cmax < np.iinfo(np.int8).max:
                df[col] = df[col].astype(np.int8)

            elif cmin > np.iinfo(np.int16).min and cmax < np.iinfo(np.int16).max:
                df[col] = df[col].astype(np.int16)
            
            elif cmin > np.iinfo(np.int32).min and cmax < np.iinfo(np.int32).max:
                df[col] = df[col].astype(np.int32)

        
        if df[col].dtypes in ["float64", "float32"]:
            
            cmin = df[col].min()
            cmax = df[col].max()
            
            if cmin > np.finfo(np.float16).min and cmax < np.finfo(np.float16).max:
                df[col] = df[col].astype(np.float16)
            
            elif cmin > np.finfo(np.float32).min and cmax < np.finfo(np.float32).max:
                df[col] = df[col].astype(np.float32)

    
    print(" ")
    print("__MEMORY USAGE AFTER COMPLETION:__")
    mem_usage = df.memory_usage().sum()/1024**2
    print("Memory Usage is :",mem_usage,"MB")
    print('This is',100*mem_usage/initial_mem_usage,'% of the initial size')

    return df
```

```python
order_products=reduce_memory(order_products)
```

### Aisles

```python
temp_df = order_products.groupby("aisle")["reordered"].agg(['count','sum']).rename(columns={'count':'total', 'sum':'reorders'})
temp_df = temp_df.sort_values('total',ascending = False).reset_index()
```

```python
temp_df
```

```python
# Total Orders and Reorders from most popular Aisles 

fig, ax = plt.subplots(figsize=(15,8))
ax = sns.barplot(y = temp_df.aisle[0:20], x= temp_df.total[0:20], color=color[0], label='total')
ax = sns.barplot(y = temp_df.aisle[0:20], x= temp_df.reorders[0:20], color=color[1], label='reordered')
ax.set_xlabel('Orders Count')
ax.set_ylabel('Aisles')
ax.set_title('Total Orders and Reorders from most popular Aisles')
```

```python
# 20 Aisles with Highest Reorder Ratio

temp_df['reorder_ratio'] = temp_df.reorders/temp_df.total
temp_df = temp_df.sort_values("reorder_ratio", ascending = False).reset_index()


fig, ax = plt.subplots(figsize=(15,8))
ax = sns.barplot(y = temp_df.aisle[0:20], x = temp_df.reorder_ratio[0:20], color = color[2])
ax.set_xlabel('Reorder Ratio')
ax.set_ylabel('Aisles')
ax.set_title('Aisles with Highest reorder Ratio')
```

```python
# 20 Aisles with lowest reorder ratio 
fig, ax = plt.subplots(figsize=(13,8))
ax = sns.barplot(y= temp_df.aisle[-20:], x= temp_df.reorder_ratio[-20:], color = color[2])
ax.set_xlabel('Reorder ratio')
ax.set_ylabel('Aisles')
ax.set_title('Aisles with lowest Reorder Ratio')
```
